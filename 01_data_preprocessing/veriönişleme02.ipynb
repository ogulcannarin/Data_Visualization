```python
import numpy as np
import pandas as pd

V1 = np.array([1, 3, 6, np.nan, 7, 1, np.nan, 9, 15])
V2 = np.array([7, np.nan, 5, 8, 12, np.nan, np.nan, 2, 3])
V3 = np.array([np.nan, 12, 5, 6, 14, 7, np.nan, 2, 31])

df = pd.DataFrame({
    "V1": V1,
    "V2": V2,
    "V3": V3
})

print(df)
```

```python
df["V1"].fillna(0,inplace=True)
df
```

```python
#Eksik değerlerin yerine sıfır değeri atama
df["V1"].fillna(df["V1"].mean(),inplace=True)#sadece V1 değişkeni için
df
```

```python
#Eksik değerlerin yerien ortalama değer ataam
df.fillna(df.mean()[:])#tüm değişkenler için
```

```python
# eksik verilerin yerine meydan atama
df.fillna(df.median(),inplace=True)
df
```

```python
import numpy as np
import pandas as pd

maas = np.array([1, 3, 6, np.nan, 7, 1, np.nan, 9, 15])
unvan = np.array(["MÜHENDİS", np.nan, "TEKNİKER", "TEKNİKER", "TEKNİKER",
                  "TEKNİKER", "TEKNİKER", "MÜHENDİS", "MÜHENDİS"], dtype=object)

df = pd.DataFrame({
    "maas": maas,
    "unvan": unvan
})

print(df)
```

```python
df.groupby("unvan")["maas"].mean()
```

```python
df["maas"].fillna(df.groupby("unvan")["maas"].transform("mean"))
```
